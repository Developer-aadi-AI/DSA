```python
class Solution:
    def canConstruct(self, ransomNote: str, magazine: str) -> bool:
        if len(ransomNote) <= len(magazine):
            counter1 = {}
            for i in magazine:
                counter1[i] = counter1.get(i, 0 ) + 1
            for i in ransomNote:
                if i in counter1.keys() and counter1[i] != 0:
                    counter1[i] -= 1
                else:
                    return False
            return True
        else:
            return False
```
