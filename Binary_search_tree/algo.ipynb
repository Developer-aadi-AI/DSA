```python
class TreeNode:
    def __init__(self, val, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

    def __str__(self):
        return str(self.val)

    
```

```python
A = TreeNode(12)
B = TreeNode(11)
C = TreeNode(13)
D = TreeNode(14)
E = TreeNode(15)
F = TreeNode(16)
G = TreeNode(17)


A.left = B
A.right = C
B.left = D
B.right = E
C.left = F
C.right = G
```

```python
#        12
#    11       13
# 14   15   16   17 
```

```python
print(A)
```

```python
# Recursive Pre-order DFS Traversal  :-  Time :- O(n), Space :- O(n)
def PreOrder(node):
    if not node:
        return
    print(node)
    PreOrder(node.left)
    PreOrder(node.right)

PreOrder(A)
```

```python
# Recursive In-order DFS Traversal :-  Time :- O(n), Space :- O(n)
def InOrder(node):
    if not node:
        return
    
    InOrder(node.left)
    print(node)
    InOrder(node.right)

InOrder(A)
```

```python
# Iterative Pre Order Traversal (DFS) Time: O(n), Space: O(n)
def post_order(node):
    if not node:
        return
        
    post_order(node.left)
    post_order(node.right)
    print(node)

post_order(A)
```
