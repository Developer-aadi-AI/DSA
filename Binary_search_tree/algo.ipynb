```python
class TreeNode:
    def __init__(self, val, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

    def __str__(self):
        return str(self.val)

    
```

```python
A = TreeNode(12)
B = TreeNode(11)
C = TreeNode(13)
D = TreeNode(14)
E = TreeNode(15)
F = TreeNode(16)
G = TreeNode(17)


A.left = B
A.right = C
B.left = D
B.right = E
C.left = F
C.right = G
```

```python
#        12
#    11       13
# 14   15   16   17 
```

```python
print(A)
```

```python
# Recursive Pre-order DFS Traversal  :-  Time :- O(n), Space :- O(n)
def PreOrder(node):
    if not node:
        return
    print(node)
    PreOrder(node.left)
    PreOrder(node.right)

PreOrder(A)
```

```python
# Recursive In-order DFS Traversal :-  Time :- O(n), Space :- O(n)
def InOrder(node):
    if not node:
        return
    
    InOrder(node.left)
    print(node)
    InOrder(node.right)

InOrder(A)
```

```python
# Recursive Post-Order Traversal (DFS) Time: O(n), Space: O(n)
def post_order(node):
    if not node:
        return
        
    post_order(node.left)
    post_order(node.right)
    print(node)

post_order(A)
```

```python
# Iterative Pre Order Traversal (DFS) Time: O(n), Space: O(n)
def Iterative_pre_order(node):
    stk = [node]
    while stk:
        node = stk.pop()
        print(node)
        if node.right: stk.append(node.right)
        if node.left: stk.append(node.left)
Iterative_pre_order(A)
```

```python
from collections import deque

def level_order(node):
    q = deque()
    q.append(node)

    while q:
        node = q.popleft()
        print(node)

        if node.left: q.append(node.left)
        if node.right: q.append(node.right)

level_order(A)
```

```python
def search(node, target):
    if not node:
        return False

    if node.val == target:
        return True

    return search(node.left, target) or search(node.right, target) 

search(node=A ,target= 12)   
    
```

```python
def bst_searching(node, target):
    if not node:
        return False

    if node.val == target:
        return True

    if node.val < target: return bst_searching(node.right, target)
    else: return bst_searching(node.left, target)
```
