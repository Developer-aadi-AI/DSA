```python
class TreeNode:
    def __init__(self, val, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

    def __str__(self):
        return str(self.val)

    
```

```python
A = TreeNode(12)
B = TreeNode(11)
C = TreeNode(13)
D = TreeNode(14)
E = TreeNode(15)
F = TreeNode(16)
G = TreeNode(17)


A.left = B
A.right = C
B.left = D
B.right = E
C.left = F
C.right = G
```

```python
print(A)
```

```python
# Traversing in Pre-order
def PreOrder(node):
    if not node:
        return
    PreOrder(node.left)
    PreOrder(node.right)
```
