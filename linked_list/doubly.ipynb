```python
# Doubly Linked lists
class DoublyList:
    def __init__(self, val, next = None, prev = None):
        self.val = val
        self.next = next
        self.prev = prev

    def __str__(self):
        return str(self.val)

    
```

```python
head = tail = DoublyList(1)
print(head)
```

```python
# Display
def display(head):
    curr = head
    elements = []
    while curr:
        elements.append(str(curr.val))
        curr = curr.next
    print(" <-> ".join(elements))

display(head)
```

```python
# insert at beginning O(1):
def insert_beginning(head, tail, val):
    new_node = DoublyList(val, next = head)
    head.prev = new_node
    return new_node, tail

head, tail = insert_beginning(head, tail, 3)

display(head)
```

```python
# insert at end O(1):
def insert_at_end(head, tail, val):
    new_node = DoublyList(val, prev = tail)
    tail.next = new_node
    return head, new_node

head, tail = insert_at_end(head, tail, 4)

display(head)
```
