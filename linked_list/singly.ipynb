```python
# Singly linked list Creation

class SinglyNode:
    def __init__(self, val, next = None):
        self.val = val
        self.next = next

    def __str__(self):
        return str(self.val)


```

```python
head = SinglyNode(1)
A = SinglyNode(3)
B = SinglyNode(5)
C = SinglyNode(7)

head.next = A
A.next = B
B.next = C

print(head)
```

```python
# Traverse the head O(n)

curr = head 
while(curr):
    print(curr)
    curr = curr.next
```

```python
# Display Linked list
def display(head):
    curr = head
    element = []
    while(curr):
        element.append(str(curr.val))
        curr = curr.next
    print(" -> ".join(element))

display(head)
```

```python
# Searching in linked list O(n)
def search(head, val):
    curr = head
    while curr:
        if val == curr.val:
            return True
        curr = curr.next
    return False

print(search(head, 8))
```
