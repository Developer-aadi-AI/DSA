```python
# Traditional Binary search :- Time O(log2 n), space O(1)
arr = [-3, -1, 0, 1, 2, 3, 5, 6, 7]

def Binary_search(arr, target):
    n = len(arr)
    l = 0
    r = n - 1

    while l <= r:
        m = l + ((r - l) // 2)
        if arr[m] == target:
            return True
        elif arr[m] < target:
            l = m + 1
        else:
            r = m - 1
    return False

Binary_search(arr, 6)
```

```python
# Based on a Condition
# [T, T, T, T, T, F, F, F, F, F]
# find the false starting index 


def bin_search(brr):
    n = len(brr)
    l = 0
    r = n - 1

    while l < r:
        m = (l + r) // 2
        if brr[m]:
            r = m
        else:
            l = m + 1

    return l
```

```python
brr = [False, False, False, False, False, False, False, False, False, False, True, True, True, True, True, True, True] # 0, 8, 12
```
