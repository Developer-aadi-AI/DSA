```python
# Traditional Binary search :- Time O(log2 n), space O(1)
arr = [-3, -1, 0, 1, 3, 5, 6, 7]

def Binary_search(arr, target):
    n = len(arr)
    l = 0
    r = n - 1

    while l <= r:
        m =l + (r - l) // 2
        if arr[m] == target:
            return True
        elif arr[m] < target:
            l = m + 1
        else:
            r = m - 1
    return False

Binary_search(arr, 5)
```
